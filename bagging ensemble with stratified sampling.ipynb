```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf 
from PIL import Image
```

```python
import os 
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from keras.models import Sequential 
from keras.layers import Conv2D, MaxPool2D, Dense, Flatten, Dropout,AveragePooling2D
```

```python
data = []
labels = []
classes = 43 
cur_path = os.getcwd() 
for i in range(classes): 
    path = os. path.join(cur_path,'train', str(i)) 
    images = os.listdir(path) 
    for a in images: 
        try: 
            image = Image.open(path + '/' + a) 
            image = image.resize((75,75)) 
            image = np.array(image) 
            data.append(image) 
            labels.append(i) 
        except: 
            print("Error loading image") 
data = np.array(data)
labels = np.array(labels)
```

```python
print(data.shape, labels.shape)
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.ensemble import BaggingClassifier

sss = StratifiedShuffleSplit(n_splits=2, test_size=0.2, random_state=42)
```

```python
# Loop through the splits
from keras.models import Sequential 
for i,(train_index, test_index) in enumerate(sss.split(data, labels)):
    # Split the dataset into training and testing sets
    X_train, X_test = data[train_index], data[test_index]
    y_train, y_test = labels[train_index], labels[test_index]
    
    y_train = to_categorical(y_train, 43)
    y_test = to_categorical(y_test, 43)
    # Define model 1
    model = Sequential()
    model.add(Conv2D(filters=32, kernel_size=(5,5), activation='relu', input_shape=X_train.shape[1:]))
    model.add(Conv2D(filters=32, kernel_size=(5,5), activation='relu'))
    model.add(MaxPool2D(pool_size=(2, 2)))
    model.add(Dropout(rate=0.25))
    model.add(Conv2D(filters=64, kernel_size=(3, 3), activation='relu'))
    model.add(Conv2D(filters=64, kernel_size=(3, 3), activation='relu'))
    model.add(MaxPool2D(pool_size=(2, 2)))
    model.add(Dropout(rate=0.25))
    model.add(Flatten())
    model.add(Dense(256, activation='relu'))
    model.add(Dropout(rate=0.5))
    model.add(Dense(43, activation='softmax'))
    model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
    model.save(f'traffic_classifier_basic_CNN_{i}.h5')
    plt.figure(0)
    plt.plot(history.history['accuracy'], label='training accuracy')
    plt.plot(history.history['val_accuracy'], label='val accuracy')
    plt.title('Accuracy')
    plt.xlabel('epochs')
    plt.ylabel('accuracy')
    plt.legend()
    plt.show()
    plt.figure(1)
    plt.plot(history.history['loss'], label='training loss')
    plt.plot(history.history['val_loss'], label='val loss')
    plt.title('Loss')
    plt.xlabel('epochs')
    plt.ylabel('loss')
    plt.legend()
    plt.show()
    # Define model 2
    model1 = Sequential()
    model1.add(Conv2D(filters=6, kernel_size=(5, 5), activation='relu', input_shape=X_train.shape[1:]))
    model1.add(AveragePooling2D(pool_size=(2, 2)))
    model1.add(Conv2D(filters=16, kernel_size=(5, 5), activation='relu'))
    model1.add(AveragePooling2D(pool_size=(2, 2)))
    model1.add(Flatten())
    model1.add(Dense(units=120, activation='relu'))
    model1.add(Dense(units=84, activation='relu'))
    model1.add(Dense(units=43, activation = 'softmax'))
    model1.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    history = model1.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
    model.save(f'traffic_classifier_LeNet_CNN_{i}.h5')
    plt.figure(0)
    plt.plot(history.history['accuracy'], label='training accuracy')
    plt.plot(history.history['val_accuracy'], label='val accuracy')
    plt.title('Accuracy')
    plt.xlabel('epochs')
    plt.ylabel('accuracy')
    plt.legend()
    plt.show()
    plt.figure(1)
    plt.plot(history.history['loss'], label='training loss')
    plt.plot(history.history['val_loss'], label='val loss')
    plt.title('Loss')
    plt.xlabel('epochs')
    plt.ylabel('loss')
    plt.legend()
    plt.show()
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
y_test = pd.read_csv('Test.csv')
labels = y_test["ClassId"].values
imgs = y_test["Path"].values
data=[]
for img in imgs:
   image = Image.open(img)
   image = image.resize((75,75))
   data.append(np.array(image))
X_test=np.array(data)
```

```python
from keras.models import load_model

model1 = load_model('traffic_classifier_basic_CNN_0.h5')
model2 = load_model('traffic_classifier_basic_CNN_1.h5')
model3 = load_model('traffic_classifier_LeNet_CNN_0.h5')
model4 = load_model('traffic_classifier_LeNet_CNN_1.h5')
```

```python
prediction1 = model1.predict(X_test)
prediction2 = model2.predict(X_test)
prediction3 = model3.predict(X_test)
prediction4 = model4.predict(X_test)
```

```python
classes_x_1=np.argmax(prediction1,axis=1)
classes_x_2=np.argmax(prediction2,axis=1)
classes_x_3=np.argmax(prediction3,axis=1)
classes_x_4=np.argmax(prediction4,axis=1)
```

```python
accuracy1 = accuracy_score(labels, classes_x_1)
accuracy2 = accuracy_score(labels, classes_x_2)
accuracy3 = accuracy_score(labels, classes_x_3)
accuracy4 = accuracy_score(labels, classes_x_4)
print('Accuracy Score for model1 = ', accuracy1)
print('Accuracy Score for model2 = ', accuracy2)
print('Accuracy Score for model3 = ', accuracy3)
print('Accuracy Score for model4 = ', accuracy4)
```

```python
models = [model1, model2, model3, model4]

preds = [model.predict(X_test) for model in models]
preds=np.array(preds)
summed = np.sum(preds, axis=0)

# argmax across classes
ensemble_prediction = np.argmax(summed, axis=1)
```

```python
ensemble_accuracy = accuracy_score(labels, ensemble_prediction)
print('Accuracy Score for average ensemble = ', ensemble_accuracy)
```
